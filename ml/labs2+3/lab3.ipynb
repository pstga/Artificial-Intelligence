```python

import numpy as np
import matplotlib.pyplot as plt

class KNN_classifier:
    def __init__(self, train_images, train_labels):
        self.train_images = train_images
        self.train_labels = train_labels

    def classify_image(self, test_image, num_neighbors=3, metric='l2'):
        if metric.lower() == 'l1':
            distances = np.sum(np.abs(self.train_images - test_image), axis=1)
        elif metric.lower() == 'l2':
            distances = np.sqrt(np.sum((self.train_images - test_image) ** 2, axis=1))
        else:
            raise ValueError("Metrica necunoscută! Folosește 'l1' sau 'l2'.")

        sorted_indices = np.argsort(distances)
        nearest_indices = sorted_indices[:num_neighbors]

        nearest_labels = self.train_labels[nearest_indices]

        votes = np.bincount(nearest_labels)
        predicted_label = np.argmax(votes)

        return predicted_label

imgs_train = np.loadtxt('data/train_images.txt')
labels_train = np.loadtxt('data/train_labels.txt').astype(int)
imgs_test = np.loadtxt('data/test_images.txt')
labels_test = np.loadtxt('data/test_labels.txt').astype(int)

knn = KNN_classifier(imgs_train, labels_train)

for k in [1, 3, 5, 7, 9]:
    correctly_classified = 0
    for i, img in enumerate(imgs_test):
        prediction = knn.classify_image(img, num_neighbors=k, metric='l2')
        if prediction == labels_test[i]:
            correctly_classified += 1

    accuracy = correctly_classified / len(imgs_test)
    print(f"Accuracy for k = {k}: {accuracy*100:.2f}%")
```
