# 1. Alpha-Beta Pruning

Aplică Alpha-Beta Pruning pe jocul X&0

```python
import copy

class Nod:
    def __init__(self, parinte=None, table=None):
        if table is None:
            self.table = [['_', '_', '_'] for _ in range(3)]
        else:
            self.table = table
        self.parinte = parinte
        self.succesori = []
        self.scor = self.euristica()
        self.scor_evaluat = None  # valoarea nodului

    def set_parinte(self, nod):
        self.parinte = nod

    def __str__(self):
        return '\n'.join('|'.join(row) for row in self.table)

    def __repr__(self):
        return f"Nod(Scor: {self.scor})"

    def scor_linie_individuala(self, linie):
        nr_X = linie.count('X')
        nr_O = linie.count('O')

        if nr_X > 0 and nr_O > 0:
            return 0

        if nr_X == 3: return 100
        if nr_X == 2: return 10
        if nr_X == 1: return 1

        if nr_O == 3: return -100
        if nr_O == 2: return -10
        if nr_O == 1: return -1

        return 0

    def euristica(self):
        scor_total = 0
        for i in range(3):
            scor_total += self.scor_linie_individuala(self.table[i])

        for j in range(3):
            coloana = [self.table[i][j] for i in range(3)]
            scor_total += self.scor_linie_individuala(coloana)

        diag_principala = [self.table[i][i] for i in range(3)]
        diag_secundara = [self.table[i][2 - i] for i in range(3)]

        scor_total += self.scor_linie_individuala(diag_principala)
        scor_total += self.scor_linie_individuala(diag_secundara)

        return scor_total

    def drumRadacina(self):
        noduri = []
        temp = self
        while temp is not None:
            noduri.append(temp)
            temp = temp.parinte
        noduri.reverse()
        return noduri

    def vizitat(self, nod) -> bool:
        for n in self.drumRadacina():
            if n.table == nod.table:
                return True
        return False

class Graf:
    def __init__(self, nodStart):
        self.nodStart = nodStart

    def succesori(self, nod, symbol):
        lista_succesori = []
        for r in range(3):
            for c in range(3):
                if nod.table[r][c] == '_':
                    noua_tabla = copy.deepcopy(nod.table)
                    noua_tabla[r][c] = symbol

                    nod_nou = Nod(parinte=nod, table=noua_tabla)
                    if not nod.vizitat(nod_nou):
                        lista_succesori.append(nod_nou)

        nod.succesori = lista_succesori
        return lista_succesori


def alpha_beta(nodCurent, jucator, alpha=-float('inf'), beta=float('inf'), adancime=10):
    # win condition
    if nodCurent.scor >= 100 or nodCurent.scor <= -100:
        return nodCurent.scor

    lista_succesori = graf.succesori(nodCurent, jucator)

    # alt wincon / remiza sau tabla plina pretty much
    if adancime == 0 or not lista_succesori:
        return nodCurent.scor

    urmatorul_jucator = 'O' if jucator == 'X' else 'X'

    if jucator == 'X':  # jucatorul MAX
        val = -float('inf')
        for copil in lista_succesori:
            val = max(val, alpha_beta(copil, urmatorul_jucator, alpha, beta, adancime - 1))
            alpha = max(alpha, val)
            if beta <= alpha:
                break  # pruning ul propriu zis
        nodCurent.scor_evaluat = val
        return val

    else:
        val = float('inf')
        for copil in lista_succesori:
            val = min(val, alpha_beta(copil, urmatorul_jucator, alpha, beta, adancime - 1))
            beta = min(beta, val)
            if beta <= alpha:
                break  # pruning ul propriu zis
        nodCurent.scor_evaluat = val
        return val

tabla_initiala = [
    ['_', '_', '_'],
    ['_', '_', '_'],
    ['_', '_', '_']
]

nod_start = Nod(table=tabla_initiala)
graf = Graf(nod_start)

rezultat = alpha_beta(graf.nodStart, jucator='X', adancime=9)
print(f"optim: {rezultat}")
```
