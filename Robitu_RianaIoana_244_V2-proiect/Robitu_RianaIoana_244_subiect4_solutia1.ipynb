```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold
from sklearn import svm

import numpy as np
from sklearn.metrics import accuracy_score

# Functia Kernel Hellinger
def hellinger_kernel(x, y):
    return 0.5 * (np.linalg.norm(x - y)**2)

def procesare_date(train_sentences_file, test_sentences_file, words_file, mapping_file):
    # incarca datele
    train_sentences = []
    with open(train_sentences_file, 'r') as f:
        for line in f:
            train_sentences.append(line.strip().split())

    test_sentences = []
    with open(test_sentences_file, 'r') as f:
        for line in f:
            test_sentences.append(line.strip().split())

    train_labels = np.load('train_coordinates.npy')
    with open(words_file, 'r') as f:
        words = [line.strip() for line in f]

    with open(mapping_file, 'r') as f:
        mapping = {}
        for line in f:
            parts = line.strip().split(',')
            if len(parts) == 2:
                char = parts[0]
                num = int(parts[1])
                mapping[char] = num
            else:
                print(f"Avertisment: linia '{line.strip()}' din {mapping_file} are un format incorect și este ignorată.")

    return train_sentences, test_sentences, train_labels, words, mapping


def genereaza_vectori(sentences, words, mapping):
    vectori = []
    for sentence in sentences:
        vector = np.zeros(len(words))
        for word in sentence:
            if word in mapping:
                vector[mapping[word]] += 1
        vectori.append(vector)
    return np.array(vectori)


def antreneaza_svm_hellinger(train_sentences, train_labels, test_sentences, words, mapping):
    train_vectors = genereaza_vectori(train_sentences, words, mapping)
    test_vectors = genereaza_vectori(test_sentences, words, mapping)

    n_train = train_vectors.shape[0]
    kernel_matrix_train = np.zeros((n_train, n_train))
    for i in range(n_train):
        for j in range(n_train):
            kernel_matrix_train[i, j] = hellinger_kernel(train_vectors[i], train_vectors[j])

    n_test = test_vectors.shape[0]
    kernel_matrix_test = np.zeros((n_test, n_test))
    for i in range(n_test):
        for j in range(n_test):
            kernel_matrix_test[i, j] = hellinger_kernel(test_vectors[i], test_vectors[j])

    # aici trebuia sa vina classifier ul knn dar NU AM RESTUL EXERCITIILOR OFFFFF

    y_pred = svm.predict(kernel_matrix_test)

    acuratete = accuracy_score(train_labels, y_pred)

    return svm, y_pred, acuratete

# mmm nasol moment
train_sentences_file = 'train_samples.txt'
test_sentences_file = 'test_samples.txt'
words_file = 'words.txt'
mapping_file = 'mapping.txt'

train_sentences, test_sentences, train_labels, words, mapping = procesare_date(train_sentences_file, test_sentences_file, words_file, mapping_file)

svm, y_pred, acuratete = antreneaza_svm_hellinger(train_sentences, train_labels, test_sentences, words, mapping)

print(f"Acuratețe pe datele de testare: {acuratete}")
```
